```python
office_status = county_pr.groupby('Employee Identifier')['Office'].agg(
    has_missing=lambda x: x.isna().any(),
    has_office=lambda x: x.notna().any()
)

employees_to_fill = office_status[
    office_status['has_missing'] & office_status['has_office']
].index
len(employees_to_fill)
```

```python
job_title_map = {
    'State\'S Attorney': 'State\'s Attorney',
    'States Attorney': 'State\'s Attorney',
    'Ambulatory Community Hlth Ntwk': 'Ambulatory/Community Health Network',
    'Ambulatory/Community Hlth Ntwk': 'Ambulatory/Community Health Network',
    'Boh  Services Core Center': 'BOH Services Core Center',
    'Boh Services Core Center': 'BOH Services Core Center',
    'Clerk Of Crct Crt Off. Of Cler': 'Clerk Of Circuit Court Office Of Clerk',
    'Clerk Of Crct Crt Off.Of Clerk': 'Clerk Of Circuit Court Office Of Clerk',
    'Warehouse Records Clerk I  Sr': 'Warehouse Records Clerk I, Sr',
    'Warehouse Records Clerk I, Sr': 'Warehouse Records Clerk I, Sr',
}

county_pr['Job Title'] = county_pr['Job Title'].replace(job_title_map)
```

```python
from difflib import get_close_matches

titles = county_pr['Job Title'].dropna().unique()

for title in titles:
    matches = get_close_matches(title, titles, n=3, cutoff=0.85)
    if len(matches) > 1:
        print(title, '->', matches[1:])
```

```python
end_date = pd.Timestamp('2018-06-30')

county_pr['Tenure Years'] = (
    (end_date - county_pr['Original Hire Date']).dt.days / 365.25
)
```

```python
county_pr['Tenure Group'] = pd.cut(
    county_pr['Tenure Years'],
    bins=[-1, 2, 5, 10, 20, float('inf')],
    labels=['0-2 Years', '2-5 Years', '5-10 Years', '10-20 Years', '20+ Years']
)
```
